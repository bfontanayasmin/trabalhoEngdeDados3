Cria os schemas Landing, Bronze, Silver e Gold no catálogo `workspace`, além do volume `workspace.landing.dados`. O volume guarda os CSVs extraídos do PostgreSQL no Neon. As demais camadas armazenam tabelas Delta.

`IF NOT EXISTS` permite repetir a preparação sem apagar os dados existentes.

```python
%sql
CREATE SCHEMA IF NOT EXISTS workspace.landing
COMMENT 'Dados brutos extraídos do banco de dados';

CREATE VOLUME IF NOT EXISTS workspace.landing.dados
COMMENT 'Arquivos CSV da loja';

CREATE SCHEMA IF NOT EXISTS workspace.bronze
COMMENT 'Dados brutos em tabelas Delta';

CREATE SCHEMA IF NOT EXISTS workspace.silver
COMMENT 'Dados tratados e validados';

CREATE SCHEMA IF NOT EXISTS workspace.gold
COMMENT 'Modelo dimensional para análise de vendas';
```

## Conferência dos schemas

Lista os schemas do catálogo para verificar a criação das quatro camadas.

```python
%sql
SHOW SCHEMAS IN workspace;
```
